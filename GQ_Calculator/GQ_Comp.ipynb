```python
import numpy as np
from matplotlib import pyplot as plt
```

$$V(\phi) = \dfrac{1}{2}V_0 \phi^2 $$
$$\Upsilon(\phi,T) = C_{\Upsilon}T $$

```python
Qp1,Gp1 = np.loadtxt("../Quad_T/p1/GQ.dat",delimiter=",",unpack=True)
Qp2,Gp2 = np.loadtxt("../Quad_T/p2/GQ.dat",delimiter=",",unpack=True)
Qp3,Gp3 = np.loadtxt("../Quad_T/p3/GQ.dat",delimiter=",",unpack=True)
Qwp3,Gwp3 = np.loadtxt("../Quad_T/p3/GQ_wi2easy.dat",unpack=True)
```

```python
plt.plot(Qp2,Gp2,label=r"$V_0 = 10^{-10} M_{Pl}^4$")
plt.plot(Qp1,Gp1,label=r"$V_0 = 10^{-14} M_{Pl}^4$",linestyle="--")
plt.plot(Qp3,Gp3,label=r"$V_0 = 10^{-18} M_{Pl}^4$",linestyle="-.")
plt.plot(Qwp3,Gwp3,label=r"WI2easy $V_0 = 10^{-18} M_{Pl}^4$",linestyle=":")
plt.xscale("log",base=10)
plt.yscale("log",base=10)
plt.xlabel("$Q$")
plt.ylabel("G(Q)")
plt.legend()
```
