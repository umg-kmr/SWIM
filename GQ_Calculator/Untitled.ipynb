```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
Qw,GQw = np.loadtxt("../../SWIM/WI2Easy/DSWIM_Exp/T3/G(Q)_data_potential=DSWIM_Exp_dissip=T3_Ne=60_noradnoise_nonthermal_BD-ICs.dat",unpack="True")
Qs,GQs = np.loadtxt("GQ_exp_stoch.dat",unpack="True",delimiter=",")
Qd,GQd = np.loadtxt("GQ_exp_det.dat",unpack="True",delimiter=",")
```

```python
plt.plot(Qw,GQw,label="WI2easy",alpha=0.8)
plt.plot(Qs,GQs,label="SWIM",linestyle="-.")
plt.plot(Qd,GQd,label="Deterministic SWIM",linestyle="--")
plt.xscale("log",base=10)
plt.yscale("log",base=10)
plt.xlabel(r"$Q$")
plt.ylabel(r"$G(Q)$")
plt.grid(alpha=0.04)
plt.axvline(350,linestyle="--",color="grey",label=r"$Q\approx 350$")
plt.axvline(2000,linestyle="--",color="grey",label=r"$Q\approx 2000$")
plt.legend()
plt.tight_layout()
plt.figure(dpi=300)
```

```python
Qw,GQw = np.loadtxt("../../SWIM/WI2Easy/DSWIM_Quartic/T3/G(Q)_data_potential=DSWIM_Quartic_dissip=T3_Ne=60_withradnoise_nonthermal_BD-ICs.dat",unpack="True")
Qs,GQs = np.loadtxt("GQ_quartic_stoch.dat",unpack="True",delimiter=",")
Qd,GQd = np.loadtxt("GQ_quartic_det.dat",unpack="True",delimiter=",")
```

```python
plt.plot(Qw,GQw,label="WI2easy",alpha=0.8)
plt.plot(Qs,GQs,label="Stochastic SWIM",linestyle="-.",alpha=0.7)
plt.plot(Qd,GQd,label="Deterministic SWIM",linestyle="--")
plt.xscale("log",base=10)
plt.yscale("log",base=10)
plt.xlabel(r"$Q$")
plt.ylabel(r"$G(Q)$")
plt.xlim(1e-2,1e1)
plt.ylim(8,1000)
plt.grid(alpha=0.04)
plt.legend()
plt.tight_layout()
plt.figure(dpi=300)
```
